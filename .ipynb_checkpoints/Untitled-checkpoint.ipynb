```python
#Method 1 : Using GoogleTranslator 
from deep_translator import GoogleTranslator

def translate_text(text, source_lang='en', target_lang='fr'):
    translated = GoogleTranslator(source=source_lang, target=target_lang).translate(text)
    return translated

# Example
print(translate_text("Hello, how are you?", "en", "ar"))
```

```python
#Method 2: Using MymemoryAPI
import requests

url = "https://api.mymemory.translated.net/get"
params = {
    "q": "Hello, how are you?",
    "langpair": "en|es"
}

response = requests.get(url, params=params)
result = response.json()
print(result["responseData"]["translatedText"])
```

```python
#Method 3: LSTM (not accurate due to small dataset)
import pandas as pd

df = pd.read_csv('data/kan.txt',sep='\t',header=None)
df=df[[0,1]]
df.columns=['English','Kannada']
print(df.head())
```

```python
def preprocessing(sentence):
    sentence = sentence.lower().strip()
    sentence = '<start>' + sentence + '<end'
    return sentence

df['English'] = df['English'].apply(preprocessing)
df['Kannada'] = df['Kannada'].apply(preprocessing)
print(df.head())
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
eng_tokenizer = Tokenizer()
eng_tokenizer.fit_on_texts(df['English'])
eng_vocab_size = len(eng_tokenizer.word_index) + 1
eng_sequences = eng_tokenizer.texts_to_sequences(df['English'])
print(eng_sequences[:5])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
eng_padded = pad_sequences(eng_sequences, padding='post')
print(eng_padded.shape)

kan_tokenizer = Tokenizer()
kan_tokenizer.fit_on_texts(df['Kannada'])
kan_vocab_size = len(kan_tokenizer.word_index) + 1
kan_sequences = kan_tokenizer.texts_to_sequences(df['Kannada'])
print(kan_sequences[:5])
kan_padded = pad_sequences(kan_sequences, padding='post')
print(kan_padded.shape)
print(kan_padded[0])
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense, Embedding

# Encoder
encoder_inputs = Input(shape=(eng_padded.shape[1],))  # Input layer for English sequences
encoder_embedding = Embedding(input_dim=eng_vocab_size, output_dim=64, mask_zero=True)(encoder_inputs)  # Embedding layer
encoder_lstm = LSTM(128, return_state=True)  # LSTM layer
encoder_outputs, state_h, state_c = encoder_lstm(encoder_embedding)  # Get hidden states
encoder_states = [state_h, state_c]  # Save states to pass to decoder

# Decoder
decoder_inputs = Input(shape=(kan_padded.shape[1],))  # Input layer for Kannada sequences
decoder_embedding = Embedding(input_dim=kan_vocab_size, output_dim=64, mask_zero=True)(decoder_inputs)  # Embedding
decoder_lstm = LSTM(128, return_sequences=True, return_state=True)  # LSTM layer
decoder_outputs, _, _ = decoder_lstm(decoder_embedding, initial_state=encoder_states)  # Connect encoder states
decoder_dense = Dense(kan_vocab_size, activation='softmax')  # Dense layer for word probabilities
decoder_outputs = decoder_dense(decoder_outputs)  # Apply dense layer

# Seq2Seq Model
model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
import numpy as np

# Prepare decoder input and output
decoder_input = kan_padded[:, :-1]   # All words except the last
decoder_target = kan_padded[:, 1:]   # All words except the first

# One-hot encode the decoder target
from tensorflow.keras.utils import to_categorical
decoder_target_one_hot = to_categorical(decoder_target, num_classes=kan_vocab_size)
print(decoder_input.shape, decoder_target_one_hot.shape)
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Embedding, Dense

# Encoder
encoder_inputs = Input(shape=(eng_padded.shape[1],))
enc_emb = Embedding(input_dim=eng_vocab_size, output_dim=256, mask_zero=True)(encoder_inputs)
encoder_lstm = LSTM(256, return_state=True)
encoder_outputs, state_h, state_c = encoder_lstm(enc_emb)
encoder_states = [state_h, state_c]

# Decoder
decoder_inputs = Input(shape=(decoder_input.shape[1],))
dec_emb = Embedding(input_dim=kan_vocab_size, output_dim=256, mask_zero=True)(decoder_inputs)
decoder_lstm = LSTM(256, return_sequences=True, return_state=True)
decoder_outputs, _, _ = decoder_lstm(dec_emb, initial_state=encoder_states)
decoder_dense = Dense(kan_vocab_size, activation='softmax')
decoder_outputs = decoder_dense(decoder_outputs)

# Model
model = Model([encoder_inputs, decoder_inputs], decoder_outputs)
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
import numpy as np
from tensorflow.keras.utils import to_categorical

# Shift decoder target by one timestep
decoder_target = kan_padded[:, 1:]   # remove the first token
decoder_input = kan_padded[:, :-1]   # remove the last token

# One-hot encode the target
decoder_target_one_hot = np.zeros((decoder_target.shape[0], decoder_target.shape[1], kan_vocab_size), dtype='float32')

for i, seq in enumerate(decoder_target):
    for t, word_id in enumerate(seq):
        if word_id != 0:  # skip padding
            decoder_target_one_hot[i, t, word_id] = 1

# Train the model
model.fit([eng_padded, decoder_input], decoder_target_one_hot, batch_size=16, epochs=50, validation_split=0.2)
```

```python
import numpy as np

def translate_sentence(sentence, eng_tokenizer, kan_tokenizer, model, max_eng_len, max_kan_len):
    # 1️⃣ Tokenize and pad the input English sentence
    seq = eng_tokenizer.texts_to_sequences([sentence])
    seq_padded = pad_sequences(seq, maxlen=max_eng_len, padding='post')
    
    # 2️⃣ Start decoder with the 'start' token (usually 1)
    decoder_input_seq = np.array([[1]])  # start token id
    
    translated_sentence = []

    for i in range(max_kan_len):
        # Predict next word
        preds = model.predict([seq_padded, decoder_input_seq], verbose=0)
        next_word_id = np.argmax(preds[0, -1, :])
        
        # Stop if end token (usually 2)
        if next_word_id == 2:
            break
        
        # Append word to translated sentence
        translated_sentence.append(kan_tokenizer.index_word[next_word_id])
        
        # Update decoder input sequence
        decoder_input_seq = np.append(decoder_input_seq, [[next_word_id]], axis=1)
    
    return ' '.join(translated_sentence)

# Example
english_sentence = "Tom woke up."
translation = translate_sentence(english_sentence, eng_tokenizer, kan_tokenizer, model, eng_padded.shape[1], kan_padded.shape[1])
print("English:", english_sentence)
print("Kannada:", translation)
```

```python
#Method 5: Using M2M100
from transformers import M2M100ForConditionalGeneration, M2M100Tokenizer

print("Loading M2M100 model... (this may take a few minutes first time)")

# Load Facebook's M2M100 model
model_name = "facebook/m2m100_418M"
tokenizer = M2M100Tokenizer.from_pretrained(model_name)
model = M2M100ForConditionalGeneration.from_pretrained(model_name)

def translate_to_kannada(text):
    # Set source language to English
    tokenizer.src_lang = "en"
    
    # Tokenize input text
    encoded = tokenizer(text, return_tensors="pt")
    
    # Generate translation to Kannada
    generated_tokens = model.generate(
        **encoded,
        forced_bos_token_id=tokenizer.get_lang_id("kn"),
        max_length=128,
        num_beams=5
    )
    
    # Decode tokens to text
    translation = tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]
    return translation

# Demo examples
english_sentences = [
    "Tom woke up.",
    "I am feeling hungry.",
    "Give me half.",
    "How are you?"
]

print("Model loaded! Starting translations...\n")

for sentence in english_sentences:
    translation = translate_to_kannada(sentence)
    print(f"English: {sentence}")
    print(f"Kannada: {translation}")
    print("---")
```

```python
# English -> Hindi with transformer
from transformers import MarianMTModel, MarianTokenizer

print("Loading smaller model (200MB)...")

# This model is English->Hindi, but it's 8x smaller for testing
model_name = "Helsinki-NLP/opus-mt-en-hi"  
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

print("✅ Model loaded!")

def translate(text):
    encoded = tokenizer(text, return_tensors="pt", padding=True)
    translated = model.generate(**encoded, max_length=50)
    return tokenizer.decode(translated[0], skip_special_tokens=True)

# Test
print(translate("I am feeling hungry"))
```
